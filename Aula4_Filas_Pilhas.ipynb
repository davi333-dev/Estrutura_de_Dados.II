**Exercício 1**

```python
class SimuladorDesfazer:
    def __init__(self):
        self.pilha = []

    def executar_acao(self, acao):
        self.pilha.append(acao)
        print(f"Ação executada: {acao}")

    def desfazer(self):
        if self.pilha:
            acao = self.pilha.pop()
            print(f"Desfazendo: {acao}")
        else:
            print("Nada para desfazer.")


# Testando
editor = SimuladorDesfazer()
editor.executar_acao("digitar 'A'")
editor.executar_acao("digitar 'B'")
editor.executar_acao("apagar")

editor.desfazer()  # desfaz "apagar"
editor.desfazer()  # desfaz "digitar 'B'"
```

**Exercício 2**

```python
class FilaImpressao:
    def __init__(self):
        self.fila = []

    def enfileirar(self, documento):
        self.fila.append(documento)
        print(f"Documento adicionado: {documento}")

    def imprimir(self):
        if self.fila:
            documento = self.fila.pop(0)
            print(f"Imprimindo: {documento}")
        else:
            print("Nenhum documento na fila.")


# Testando
impressora = FilaImpressao()
impressora.enfileirar("relatorio.pdf")
impressora.enfileirar("contrato.docx")
impressora.enfileirar("foto.png")

impressora.imprimir()  # imprime "relatorio.pdf" (o mais antigo)
impressora.imprimir()  # imprime "contrato.docx"
```
